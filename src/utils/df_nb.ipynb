```python
Path_Data = '/Users/colinminini/Desktop/SCOC_ICE/LLM6G-Repository/data/'

#df = pd.read_csv(Path_Data + 'data_original.csv')
df = pd.read_csv(Path_Data + 'data_instant.csv')
i = np.random.randint(0, len(df)-5)
display(df[i:i+5])
```

```python
display(df.describe())
```

```python
import matplotlib.pyplot as plt

feature = np.random.choice(df.columns[1:])
x_values = pd.to_datetime(df['timestamp'], errors='coerce')
x = x_values if x_values.notna().any() else df['timestamp']

fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(x, df[feature], linewidth=1)
ax.set_title(f'Full timeseries for {feature}')
ax.set_xlabel('timestamp')
ax.set_ylabel('traffic (mbps)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
```
